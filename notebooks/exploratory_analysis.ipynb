```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\data\processed_data\diabetes_clean.csv")


# Pie Chart: Diabetes Distribution

counts = df["Outcome"].value_counts()
labels = ["No Diabetes", "Diabetes"]

plt.figure(figsize=(6,6))

plt.pie(
    counts,
    labels=labels,
    autopct="%1.1f%%",
    colors=["skyblue", "salmon"],
    startangle=90
)

plt.title("Proportion of Diabetes in Dataset")

plt.savefig(
    r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\results\figures\diabetes_piechart.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



# Glucose Distribution

plt.figure(figsize=(6,4))

sns.histplot(df["Glucose"], bins=30, kde=True)

plt.title("Distribution of Glucose Levels")

plt.savefig(
    r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\results\figures\glucose_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



# BMI vs Diabetes

plt.figure(figsize=(6,4))

sns.boxplot(x="Outcome", y="BMI", data=df)

plt.title("BMI Distribution by Diabetes Outcome")

plt.savefig(
    r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\results\figures\bmi_vs_diabetes.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



# Age vs Diabetes

plt.figure(figsize=(6,4))

sns.boxplot(x="Outcome", y="Age", data=df)

plt.title("Age Distribution by Diabetes Outcome")

plt.savefig(
    r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\results\figures\age_vs_diabetes.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()



# Correlation Heatmap

plt.figure(figsize=(8,6))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.savefig(
    r"C:\Users\Pavani\Desktop\diabetes-risk-assessment\results\figures\correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
